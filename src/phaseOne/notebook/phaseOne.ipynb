LIBRERIE

```python

from src.phaseOne.dataset import ImageDataset, TxtDataset
from src.phaseOne.qdrantHandler import QdrantHandler
from src.phaseOne.createEmbeddings import TextCreateEmbeddings, ImageCreateEmbeddings
from paths import TXT_DIR, IMG_DIR
from dotenv import load_dotenv
import os
```

CREAZIONE DATASET

```python
load_dotenv()
EMBEDDING_SIZE=512
```

```python
txt_dataset=TxtDataset(TXT_DIR)
```

```python
img_dataset=ImageDataset(directorypath=IMG_DIR)
```

CONNESSIONE QDRANT

```python
qdrant_handler_text=QdrantHandler(EMBEDDING_SIZE,collection_name=os.getenv("TEXT_COLLECTION_NAME"))
```

```python
qdrant_handler_img=QdrantHandler(EMBEDDING_SIZE,collection_name=os.getenv("IMAGE_COLLECTION_NAME"))
```

CREAZIONE EMBEDDINGS TESTO

```python
txt_create_embeddings=TextCreateEmbeddings(txt_dataset)
text_embeddings_object_list=txt_create_embeddings.create_embeddings()
```

CREAZIONE EMBEDDINGS IMMAGINI

```python
img_create_embeddings=ImageCreateEmbeddings(img_dataset)
img_embeddings_object_list= img_create_embeddings.create_embeddings()
```

SALVATAGGIO SU QDRANT

```python
for item in text_embeddings_object_list:
    embedding, payload= item.get_qdrant_embedding()
    qdrant_handler_text.insert_vector(vector=embedding,payload=payload)
```

```python
for item in img_embeddings_object_list:
    embedding, payload= item.get_qdrant_embedding()
    qdrant_handler_img.insert_vector(vector=embedding,payload=payload)
```
